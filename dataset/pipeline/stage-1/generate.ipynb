```python
# Configure
from pathlib import Path
from runpy import run_path

TIER = "M2"  # M2 uses all 50 BTC CRM profiles.
BLUEPRINT_COUNT = None  # Both tiers default to 180. Override with at least 175 for M1 or 150 for M2.
M1_BLUEPRINTS = None  # Use the M1 blueprints bundled in the M2 input ZIP.
INPUT_ROOT = Path("/kaggle/input")
OUTPUT = Path("/kaggle/working/output")
```

```python
# Load the attached Stage 1 input dataset
generators = list(INPUT_ROOT.rglob("stage-1/src/generate.py"))
if len(generators) != 1:
    raise ValueError("Attach one Stage 1 dataset with the input ZIP extracted.")
generator = run_path(str(generators[0]))
if "DEFAULT_BLUEPRINT_COUNTS" not in generator:
    raise ValueError("Attach the refreshed Stage 1 input dataset for the selected tier.")
```

```python
# Generate, validate, and save the blueprints and customer registry
options = dict(blueprint_count=BLUEPRINT_COUNT)
if TIER == "M2":
    options["m1_blueprints"] = Path(M1_BLUEPRINTS) if M1_BLUEPRINTS else None
registry, blueprints, sessions = generator["build_dataset"](TIER, **options)
OUTPUT.mkdir(parents=True, exist_ok=True)
generator["write_dataset"](OUTPUT, registry, blueprints)
print(f"{TIER}: {len(registry)} customers, {len(blueprints)} journeys, {sessions} sessions in {OUTPUT}")
```
