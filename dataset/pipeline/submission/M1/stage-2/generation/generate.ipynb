```python
# Configure · Attach Qwen 3 GGUF, select GPU T4 ×2, and enable Internet.
INPUT_DIR = ""  # Optional path to the extracted Stage 2 dataset.
MODEL_PATH = ""  # Optional path to the attached Qwen GGUF file.
BATCH_SIZE = 2
JOURNEYS = 5  # Use None for all journeys.
OUTPUT_DIR = "/kaggle/working/output"
```

```python
# Load · Find the attached Stage 2 bundle and read its inputs.
from pathlib import Path
from runpy import run_path

root = Path(INPUT_DIR or "/kaggle/input")
helpers = list(root.rglob("src/generation.py"))
if len(helpers) != 1:
    raise ValueError("Attach one Stage 2 input dataset or set INPUT_DIR.")
helper = helpers[0]
generator = run_path(str(helper))
inputs = generator["load_inputs"](helper.parents[1] / "input" / "stage-2-input")
```

```python
# Generate · Send complete journeys in batches and check their format.
generator["generate"](*inputs, batch_size=BATCH_SIZE, journeys=JOURNEYS,
                      output=Path(OUTPUT_DIR), model=Path(MODEL_PATH) if MODEL_PATH else None)
```
